```python
def get_next_state(state, move):
    state = list(state)
    pos = state.index(0)

    if move == "UP" and pos >= 3:
        new_pos = pos - 3
    elif move == "DOWN" and pos < 6:
        new_pos = pos + 3
    elif move == "LEFT" and pos % 3 != 0:
        new_pos = pos - 1
    elif move == "RIGHT" and pos % 3 != 2:
        new_pos = pos + 1
    else:
        return None

    state[pos], state[new_pos] = state[new_pos], state[pos]

    return tuple(state)


def dfs(state, goal, path, visited):

    if state == goal:
        return path

    visited.add(state)

    for move in ["UP", "DOWN", "LEFT", "RIGHT"]:

        next_state = get_next_state(state, move)

        if next_state is not None and next_state not in visited:

            result = dfs(next_state, goal, path + [move], visited)

            if result is not None:
                return result

    return None


# Input
initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

# Run DFS
result = dfs(initial, goal, [], set())

print("DFS Solution:")
print(result)
```

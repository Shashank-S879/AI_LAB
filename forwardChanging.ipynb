```python
# Forward Chaining Algorithm

# Initial facts
facts = {"A", "B"}

# Rules: (conditions, conclusion)
rules = [
    ({"A", "B"}, "C"),
    ({"C"}, "D"),
    ({"D"}, "E"),
    ({"A", "E"}, "F")
]

def forward_chaining(facts, rules, query):

    inferred = set(facts)

    while True:
        new_fact_added = False

        for conditions, conclusion in rules:

            if conditions.issubset(inferred):

                if conclusion not in inferred:
                    inferred.add(conclusion)
                    print("Derived:", conclusion)
                    new_fact_added = True

        if not new_fact_added:
            break

    print("\nFinal Knowledge Base:", inferred)

    if query in inferred:
        print("Query", query, "is TRUE")
    else:
        print("Query", query, "is FALSE")


# Query
query = "F"

forward_chaining(facts, rules, query)
```

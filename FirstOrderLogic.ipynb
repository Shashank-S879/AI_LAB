```python

# Knowledge Base
KB = {
    "C1": "~food(x) | likes(John,x)",
    "C2": "food(Apple)",
    "C3": "food(Vegetables)",
    "C4": "~eats(y,z) | killed(y) | food(z)",
    "C5": "eats(Anil,Peanuts)",
    "C6": "alive(Anil)",
    "C7": "~eats(Anil,w) | eats(Harry,w)",
    "C8": "killed(g) | alive(g)",
    "C9": "~alive(k) | ~killed(k)"
}

query = "likes(John,Peanuts)"

print("=" * 60)
print("             RESOLUTION INFERENCE")
print("=" * 60)

print("\nKNOWLEDGE BASE:")
for key, value in KB.items():
    print(f"{key}: {value}")

print("\nQuery:", query)
print("Negated Query: ~" + query)

# Resolution proof using unification
steps = [
    (
        "~likes(John,Peanuts)",
        "~food(x) | likes(John,x)",
        "~food(Peanuts)",
        "x = Peanuts"
    ),
    (
        "~food(Peanuts)",
        "~eats(y,z) | killed(y) | food(z)",
        "~eats(y,Peanuts) | killed(y)",
        "z = Peanuts"
    ),
    (
        "~eats(y,Peanuts) | killed(y)",
        "eats(Anil,Peanuts)",
        "killed(Anil)",
        "y = Anil"
    ),
    (
        "killed(Anil)",
        "~alive(k) | ~killed(k)",
        "~alive(Anil)",
        "k = Anil"
    ),
    (
        "~alive(Anil)",
        "alive(Anil)",
        "{}",
        "Complementary literals resolved"
    )
]

print("\nRESOLUTION STEPS:")

for i, (left, right, result, substitution) in enumerate(steps, 1):
    print(f"\nStep {i}:")
    print(" Clause 1 :", left)
    print(" Clause 2 :", right)
    print(" Substitution:", substitution)
    print(" Resolvent:", result)

print("\n" + "=" * 60)
print("             RESOLUTION PROOF TREE")
print("=" * 60)

for i, (left, right, result, substitution) in enumerate(steps, 1):
    print(f"\nStep {i}")
    print(f" {left}")
    print(f" {right}")
    print(" " + "-" * 35)
    print(f" {result}")
    print(f" [Substitution: {substitution}]")

print("\n" + "=" * 60)

# Check whether the empty clause is obtained
if steps[-1][2] == "{}":
    print("EMPTY CLAUSE OBTAINED!")
    print("\nRESULT: John likes Peanuts.")
    print("Hence Proved by Resolution.")
else:
    print("Proof unsuccessful.")
```

```python
# Vacuum Cleaner Agent

print("===== VACUUM CLEANER AGENT =====")

# Get initial room status
room = {}

room["A"] = input("Enter status of Room A (DIRTY/CLEAN): ").upper()
room["B"] = input("Enter status of Room B (DIRTY/CLEAN): ").upper()

# Get starting location
location = input("Enter current vacuum location (A/B): ").upper()

# Validate input
if room["A"] not in ["DIRTY", "CLEAN"] or room["B"] not in ["DIRTY", "CLEAN"]:
    print("Invalid room status!")
elif location not in ["A", "B"]:
    print("Invalid location!")
else:

    print("\nInitial Room Status:")
    print("Room A:", room["A"])
    print("Room B:", room["B"])
    print("Vacuum Location:", location)

    # Continue until both rooms are clean
    while True:

        print("\n-------------------------")
        print("Current Location:", location)
        print("Room Status:", room[location])

        # Check current room
        if room[location] == "DIRTY":
            print("Room is DIRTY.")
            print("Performing SUCK operation...")
            room[location] = "CLEAN"
            print("Room is now CLEAN.")

        else:
            print("Room is already CLEAN.")

        # Check whether both rooms are clean
        if room["A"] == "CLEAN" and room["B"] == "CLEAN":
            print("\n=========================")
            print("Room A: CLEAN")
            print("Room B: CLEAN")
            print("GOAL ACHIEVED!")
            print("Vacuum Cleaner Stopped.")
            print("=========================")
            break

        # Move to the other room
        if location == "A":
            location = "B"
        else:
            location = "A"

        print("Moving to Room", location)
```

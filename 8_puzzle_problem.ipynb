```python
import heapq

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


MOVES = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}



def misplaced_tiles(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != GOAL[i]:
            count += 1

    return count



def manhattan_distance(state):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            current_row = i // 3
            current_col = i % 3

            goal_pos = GOAL.index(state[i])
            goal_row = goal_pos // 3
            goal_col = goal_pos % 3

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()



def a_star(start, heuristic):


    open_list = []

    h = heuristic(start)
    heapq.heappush(open_list, (h, 0, start, []))

    visited = set()

    while open_list:

        f, g, state, path = heapq.heappop(open_list)

        if state in visited:
            continue

        visited.add(state)

        h = f - g

        print("State:")
        print_puzzle(state)
        print("g =", g)
        print("h =", h)
        print("f =", f)
        print("----------------")


        if state == GOAL:
            print("GOAL REACHED!")
            print("Solution moves:", path)
            print("Number of moves:", g)
            return


        blank = state.index(0)

        for new_pos in MOVES[blank]:

            new_state = list(state)


            new_state[blank], new_state[new_pos] = \
                new_state[new_pos], new_state[blank]

            new_state = tuple(new_state)

            if new_state not in visited:

                new_g = g + 1
                new_h = heuristic(new_state)
                new_f = new_g + new_h

                heapq.heappush(
                    open_list,
                    (new_f, new_g, new_state, path + [new_pos])
                )


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)



print("===== CASE 1: MISPLACED TILES =====")

a_star(start, misplaced_tiles)


print("\n===== CASE 2: MANHATTAN DISTANCE =====")

a_star(start, manhattan_distance)
```

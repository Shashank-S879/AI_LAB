```python
# Unification Algorithm

def is_variable(x):
    return x.islower()

def unify(x, y, substitution):
    if substitution is None:
        return None

    if x == y:
        return substitution

    if is_variable(x):
        substitution[x] = y
        return substitution

    if is_variable(y):
        substitution[y] = x
        return substitution

    return None


def unification(expr1, expr2):
    predicate1, args1 = expr1
    predicate2, args2 = expr2

    if predicate1 != predicate2:
        return None

    if len(args1) != len(args2):
        return None

    substitution = {}

    for x, y in zip(args1, args2):
        substitution = unify(x, y, substitution)

        if substitution is None:
            return None

    return substitution


# Example
E1 = ("Eats", ["x", "Apple"])
E2 = ("Eats", ["Riya", "y"])

result = unification(E1, E2)

if result is not None:
    print("Unification Successful")
    print("MGU:", result)
else:
    print("Unification Failed")
```

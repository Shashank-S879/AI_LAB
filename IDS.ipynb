```python
def get_next_state(state, move):
    state = list(state)
    pos = state.index(0)

    if move == "UP" and pos >= 3:
        new_pos = pos - 3
    elif move == "DOWN" and pos < 6:
        new_pos = pos + 3
    elif move == "LEFT" and pos % 3 != 0:
        new_pos = pos - 1
    elif move == "RIGHT" and pos % 3 != 2:
        new_pos = pos + 1
    else:
        return None

    state[pos], state[new_pos] = state[new_pos], state[pos]

    return tuple(state)


def dls(state, goal, limit, path):

    if state == goal:
        return path

    if limit == 0:
        return None

    for move in ["UP", "DOWN", "LEFT", "RIGHT"]:

        next_state = get_next_state(state, move)

        if next_state is not None and next_state not in path:

            result = dls(
                next_state,
                goal,
                limit - 1,
                path + [next_state]
            )

            if result is not None:
                return result

    return None


def ids(initial, goal):

    depth = 0

    while True:

        result = dls(initial, goal, depth, [initial])

        if result is not None:
            return result

        depth += 1


# Input
initial = (1, 2, 3,
           4, 0, 6,
           7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

# Run IDS
result = ids(initial, goal)

print("IDS Solution:")
print(result)
```
